```python
headers = {"Authorization": f"Bearer {api_key}",
            "Content-Type": "application/json"}

messages = [
    {
        "role": "system",
        "content": "You are a helpful assistant. Answer clearly and concisely.",
    }
]

def ask(user_message):
    messages.append({
        "role": "user",
        "content": user_message})

    payload = {
        "model": "deepseek-v4-pro",
        "messages": messages}

    response = requests.post(url, headers=headers, json=payload, timeout=60)
    response.raise_for_status()

    data = response.json()
    assistant_message = data["choices"][0]["message"]

    messages.append(assistant_message)
    return assistant_message["content"]
```

```python
response.json()#%%



# ─────────────────────────── 1. Configuration object ────────────────────────────
@dataclass(slots=True)
class OllamaClassifierConfig:
    label_map: dict[int, str]
    system: str
    fewshot: str
    model_name: str = "llama3.1:70b"
    url: str = "http://localhost:11434/api/generate"
    temperature: float = 0.0
    timeout: int = 60
    num_predict: int = 3
    stream: bool = False
    extra_options: dict[str, Any] = field(default_factory=dict)

    @property
    def label2id(self) -> dict[str, int]:
        return {v: k for k, v in self.label_map.items()}

    def build_payload(self, passage: str) -> dict[str, Any]:
        options = {
            "temperature": self.temperature,
            "num_predict": self.num_predict,
            **self.extra_options,
        }
        return {
            "model": self.model_name,
            "system": self.system,
            "prompt": self.fewshot.format(passage.strip()),
            "options": options,
            "stream": self.stream,
        }


# ─────────────────────────── 2. Single classification ────────────────────────────
def classify_single(
    passage: str,
    config: OllamaClassifierConfig,
    session: requests.Session | None = None,
) -> int | None:
    """Return a numeric label ID or None on failure."""
    if passage is None or pd.isna(passage):
        return None

    payload = config.build_payload(passage)
    http = session or requests

    try:
        resp = http.post(config.url, json=payload, timeout=config.timeout)
    except requests.RequestException as e:
        print("⛔ LLM request failed:", e)
        return None

    if resp.status_code != 200:
        print(f"⛔ LLM HTTP {resp.status_code}: {resp.text[:200]}")
        return None

    try:
        data = resp.json()
    except ValueError:
        print("⚠︎ Invalid JSON response:", resp.text[:200])
        return None

    raw = str(data.get("response", "")).strip()

    try:
        pred_id = int(raw)
    except ValueError:
        print("⚠︎ Unexpected LLM output:", data)
        return None

    return pred_id if pred_id in config.label_map else None


# ─────────────────────────── 3. Batch classification ────────────────────────────
def classify_many(
    passages,
    config: OllamaClassifierConfig,
    max_workers: int = 16,
) -> list[int | None]:
    results: list[int | None] = [None] * len(passages)

    def _task(passage: str) -> int | None:
        with requests.Session() as session:
            return classify_single(passage, config=config, session=session)

    with ThreadPoolExecutor(max_workers=max_workers) as ex:
        fut_to_i = {
            ex.submit(_task, passage): i
            for i, passage in enumerate(passages)
        }
        for fut in as_completed(fut_to_i):
            i = fut_to_i[fut]
            results[i] = fut.result()

    return results
```

```python
def raw_with_lemmata(row):
    raw_text = row["kwic_text"]
    lemmata = " ".join([t["lemma"] for t in row["kwic_tokens"] if t["pos"] in ["NOUN", "ADJ", "VERB", "PROPN"]])
    return "raw text of the passage: {0} ; filtered lemmata from the passage: {1}".format(raw_text, lemmata)
```

```python
passio_sample = passio_df.sample(10).copy()
passio_sample["passage"] = passio_sample.apply(raw_with_lemmata, axis=1)
passio_sample = passio_sample[["passage"]]
```

```python

config = OllamaClassifierConfig(
    label_map=llmprompt_passio.LABEL_MAP,
    system=llmprompt_passio.SYSTEM,
    fewshot=llmprompt_passio.FEWSHOT,
    model_name="llama3.1:70b",
    temperature=0.0,
    timeout=60,
)
passio_sample["class_llama3.1"] = classify_many(passio_sample["passage"], config)
```
