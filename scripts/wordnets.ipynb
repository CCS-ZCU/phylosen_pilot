```python
import requests
import json
import pandas as pd
```

```python
import urllib3
urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)  # suppresses the warning
```

```python
latin_lexeme = "passio"
pos = ""
base_url_latin = "https://latinwordnet.exeter.ac.uk/api/lemmas/{0}/{1}synsets/"
```

```python
resp_json = requests.get(base_url_latin.format(latin_lexeme, pos), verify=False).json()
resp_json
```

```python
greek_lexeme = "πάθος"
pos = "" # n/
base_url_greek = "https://greekwordnet.chs.harvard.edu/api/lemmas/{0}/{1}synsets"
resp_json = requests.get(base_url_greek.format(greek_lexeme, pos), verify=False).json()
```
