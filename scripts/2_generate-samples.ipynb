```python
outdir = Path("../data/large_files/phylosen_instances/")
```

```python
def read_hits(fname):
    try:
        path = os.path.join(outdir, fname)
        instances = pd.read_parquet(path)
    except:
        instances = pd.DataFrame()
    return instances
```

```python
instances = read_hits("seminarium.parquet")
len(instances)
```

```python
sample_size = 2000

for term in ["passio", "pathos"]:
    instances = pd.concat([instances, read_hits(f"{term}.parquet")], ignore_index=True)
    if len(instances) > sample_size:
        instances = instances.sample(n=2000, random_state=42).reset_index(drop=True)
    instances.to_parquet("../data/large_files/phylosen_instances/{0}_sample{1}.parquet".format(term, sample_size), index=False)
```
