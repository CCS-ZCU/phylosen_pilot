```python
import matplotlib.pyplot as plt

GROUP_COLORS = {
    "HiG":   "#C65D3B",
    "CompG": "#16858C",
    "CogG":  "#7356A8",
}


def expertise_pie(
    hig,
    compg,
    cogg,
    filename="expertise.png",
    size=5,
    dpi=300,
):
    """
    Create a tiny transparent expertise pie chart.

    Parameters
    ----------
    hig, compg, cogg : float
        Relative contributions. They do not have to sum to 100.
        Examples: (80, 20, 0), (60, 20, 20), (1, 1, 1).
    filename : str
        Output PNG filename.
    size : float
        Figure size in inches.
    dpi : int
        Output resolution.
    """

    values = [hig, compg, cogg]

    if sum(values) <= 0:
        raise ValueError("At least one contribution must be > 0.")
    fig, ax = plt.subplots(figsize=(size, size), dpi=dpi)

    # Explicitly transparent figure and axes

    fig.patch.set_alpha(0)

    ax.patch.set_alpha(0)

    ax.pie(

        values,

        colors=[

            GROUP_COLORS["HiG"],

            GROUP_COLORS["CompG"],

            GROUP_COLORS["CogG"],

        ],

        radius=1.15,          # <-- actual circle size

        startangle=90,

        counterclock=False,

        wedgeprops={"linewidth": 0},

    )

    ax.set_aspect("equal")

    ax.axis("off")
        #plt.savefig(
        #  filename,
        #  dpi=dpi,
        #    transparent=True,
        #    bbox_inches="tight",
        #    pad_inches=0,
        #)
        #plt.close(fig)
```

```python
expertise_pie(30, 20, 50, filename="expertise.png", size=0.8, dpi=300)
```
